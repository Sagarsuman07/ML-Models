## Data Cleaning

```python
# Currently the price include dollar symbol and comma like this $10,300 which we have to filter

df["price"]=(df["price"].str.replace("$","",regex=False)
            .str.replace(",","",regex=False)
            .astype(int))

df["price"]
```

```python
# Handle Milage same as price (Milage is in this form 51,000 mi.)

df["milage"]=(df["milage"].str.replace("mi.","",regex=False)
             .str.replace(",","",regex=False)
             .astype(int))
df["milage"].describe()
```

```python
df.info()
```

##### Handling Missing Values #######

```python
df.isnull().sum()
```

```python
# Handling missing values of "fuel_type"
# Since fuel type is of type object we will use mode to impute

df["fuel_type"]=df["fuel_type"].fillna(df["fuel_type"].mode()[0])
df["fuel_type"].unique()
```

```python
#Handling "accident" feature and "clean_title" feature

df["accident"]=df["accident"].fillna("Unknown")
df["clean_title"]=df["clean_title"].fillna("Unknown")
```

```python
print(df["accident"].unique())
print(df["clean_title"].unique())
```

```python

#Check for duplicate rows
df.duplicated().sum()
```

## Feature Engineering

```python
# Instead of "model_year" we can create feature car age

curr_yr=2026
df["car_age"]=curr_yr-df["model_year"]

df.drop("model_year", axis=1, inplace=True)

df["car_age"].describe()
```

```python
df["accident"].unique()
```

```python
# Now we will change these values [None reported,At least 1 accident,Unknown] into [No Accident,Accident,Unknown]


# Create a dictionary mapping the old names to the new names
accident_mapping={
    "None reported": "No Accident",
    "At least 1 accident or damage reported": "Accident",
    "Unknown": "Unknown"
}

# Replace old value to new
df["accident"]=df["accident"].replace(accident_mapping)
```

```python
df["accident"].unique()
```

```python

# Clean the title, convert ["Yes","Unknown"] to [0,1]

df["clean_title"]=(df["clean_title"].replace("Yes",1).replace("Unknown",0).astype(int))
```

```python
df["clean_title"].dtype
```

#### Creating new feature "is premium" for premium brand cars

```python
#Check all the brands name
df["brand"].value_counts()
```

```python
# 1. Define the set of luxury and exotic brands
luxury_brands = {
    "BMW", "Mercedes-Benz", "Porsche", "Audi", "Lexus", "Land", 
    "Cadillac", "Tesla", "Acura", "INFINITI", "Lincoln", "Jaguar", 
    "Volvo", "Maserati", "Bentley", "Lamborghini", "Genesis", "Alfa", 
    "Rivian", "Ferrari", "Rolls-Royce", "Aston", "McLaren", "Lotus", 
    "Lucid", "Karma", "Saab", "Bugatti", "Polestar", "Maybach"
}

# 2. Create the binary column (1 for luxury, 0 for mainstream)
df["is_luxury"]=df["brand"].isin(luxury_brands).astype(int)

df["is_luxury"].value_counts()
```

#### Create feature engine size from engine

```python
# Every engine include engine size, so we can extract them and create new feature

# Extract engine size (in liters)
df["engine_size"] = df["engine"].str.extract(r'(\d+\.?\d*)\s*(?:L|Liter)', expand=False)

# Convert to float
df["engine_size"] = df["engine_size"].astype(float)

df[["engine", "engine_size"]].head()
```
